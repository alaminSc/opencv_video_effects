# Scrolling Text Animation

```python
import cv2
rain = cv2.VideoCapture('rain.mp4')
y = 400
while True:
    ref,frame = rain.read()
    if not ref:
        break
    farme = cv2.resize(frame,(600,400))
    blur = cv2.GaussianBlur(farme,(5,5),0)
    _, thresh = cv2.threshold(blur, 127, 255, cv2.THRESH_TRUNC)
    cv2.putText(thresh,"Designed By",(265, y-30),cv2.FONT_HERSHEY_SCRIPT_SIMPLEX,0.7,(255, 255, 255),1)
    cv2.putText(thresh,"Alamin Elkhir",(250, y),cv2.FONT_HERSHEY_SCRIPT_SIMPLEX,1,(255, 255, 255),1)
    cv2.imshow("rain",thresh)
    y -=3
    if y < -50:
        y = 400
    y -=3
    if cv2.waitKey(50) & 0xFF ==27:
        break
rain.release()
cv2.destroyAllWindows()
```

# Digital Glitch Effect with OpenCV

```python
import cv2
import numpy as np
import random
cap = cv2.VideoCapture("rain.mp4")
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.resize(frame,(700,500))
    h, w = frame.shape[:2]
    glitch = frame.copy()
    for i in range(16):
        y = random.randint(0, h - 20)
        height = random.randint(5, 30)
        shift = random.randint(-50, 50)
        y2 = min(y + height, h)
        if shift > 0:
            glitch[y:y2, shift:w] = frame[y:y2, 0:w-shift]
        else:
            shift = abs(shift)
            glitch[y:y2, 0:w-shift] = frame[y:y2, shift:w]
    cv2.imshow("Glitch Effect", glitch)
    if cv2.waitKey(30) & 0xFF == 27:
        break
cap.release()
cv2.destroyAllWindows()
```

# Cartoun & Canny Effect

```python
import cv2
import numpy as np
cap = cv2.VideoCapture("slow_traffic_small.mp4")
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.resize(frame, (600, 400))
    color = cv2.bilateralFilter(frame, 9, 150, 150)
    hsv = cv2.cvtColor(color, cv2.COLOR_BGR2HSV)
    hsv[:, :, 1] = np.clip(hsv[:, :, 1] * 1.5, 0, 255)
    hsv[:, :, 2] = np.clip(hsv[:, :, 2] * 1.1, 0, 255)
    color = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (5, 5), 0)
    edges = cv2.Canny(gray,50,150)
    kernel = np.ones((2, 2), np.uint8)
    edges = cv2.dilate(edges, kernel, iterations=1)

    edges = cv2.bitwise_not(edges)

    cartoon = cv2.bitwise_and(color,color,mask=edges)
    
    cv2.imshow("Cartoon + Canny", cartoon)

    if cv2.waitKey(30) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
